```python
def con(x):
    pre='0'
    if( int(x)<=99) and (int(x)>=10) :
        return pre+str(x)
    elif int(x)<10:
        return pre+pre+str(x)
    else:
        return str(x)

df_cards['cvv']=df_cards['cvv'].apply(lambda x : con(x)) 
```

4. Exploratory Data Analysis (EDA)

Grouping & aggregation:

Pattern exploration:

Income vs. debt analysis.

Card usage distribution across clients.

Correlation matrix with corr().

Key checks:
